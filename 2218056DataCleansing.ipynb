# 15.Membuat Salinan Dataset

Dataset asli tidak langsung diubah. Membuat salinan bernama df_clean sebagai dataset hasil cleansing.

```python
df_clean = df.copy()

print("DataFrame cleansing berhasil dibuat.")
```

# 16.Membersihkan Spasi pada Data Teks

Menghapus spasi berlebih pada awal dan akhir teks, misalnya " Columbus " menjadi "Columbus".

```python
object_columns = df_clean.select_dtypes(include='object').columns

for col in object_columns:
    df_clean[col] = (
        df_clean[col]
        .astype('string')
        .str.strip()
    )

print("Spasi berlebih berhasil dibersihkan.")
```

# 17.Membersihkan Kolom Customer Name

Menyeragamkan penulisan nama pelanggan agar lebih rapi.

```python
df_clean['Customer Name'] = (
    df_clean['Customer Name']
    .astype('string')
    .str.strip()
    .str.title()
)

print(df_clean['Customer Name'].head())
```

# 18.Membersihkan Kolom Sentiment

Pada dataset terdapat beberapa bentuk penulisan sentiment seperti POSITIVE, positive, Positiv, Netral, VeryNegative, dan sebagainya. Semuanya dibuat menjadi kategori yang seragam.

```python
sentiment_key = (
    df_clean['Sentiment']
    .astype('string')
    .str.lower()
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)

sentiment_map = {
    'positive': 'Positive',
    'positiv': 'Positive',
    'very positive': 'Very Positive',

    'negative': 'Negative',
    'negatif': 'Negative',
    'very negative': 'Very Negative',
    'verynegative': 'Very Negative',

    'neutral': 'Neutral',
    'netral': 'Neutral'
}

df_clean['Sentiment'] = sentiment_key.map(sentiment_map)

print(df_clean['Sentiment'].value_counts(dropna=False))
```

# 19.Membersihkan Kolom Reason

Menyamakan variasi penulisan kategori alasan pelanggan.

```python
reason_key = (
    df_clean['Reason']
    .astype('string')
    .str.lower()
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)

reason_map = {
    'billing question': 'Billing Question',
    'billing questions': 'Billing Question',
    'payment': 'Payment',
    'payments': 'Payment',
    'service outage': 'Service Outage'
}

df_clean['Reason'] = reason_key.map(reason_map)

print(df_clean['Reason'].value_counts(dropna=False))
```

# 20.Membersihkan Kolom City

Membersihkan spasi, menyeragamkan huruf, serta mengubah singkatan kota tertentu menjadi nama yang lebih konsisten.

```python
city_key = (
    df_clean['City']
    .astype('string')
    .str.strip()
    .str.lower()
)

city_map = {
    'ft. wayne': 'Fort Wayne',
    'ft. pierce': 'Fort Pierce',
    'ft. worth': 'Fort Worth'
}

df_clean['City'] = (
    city_key
    .map(city_map)
    .fillna(city_key.str.title())
)

print(df_clean['City'].value_counts(dropna=False))
```

# 21.Membersihkan Kolom State

Kolom State memiliki campuran antara singkatan seperti CA, TX, MI dan nama lengkap. Semuanya diubah menjadi nama lengkap.

```python
state_key = (
    df_clean['State']
    .astype('string')
    .str.strip()
    .str.lower()
)

state_map = {
    'al': 'Alabama',
    'ar': 'Arkansas',
    'az': 'Arizona',
    'ca': 'California',
    'co': 'Colorado',
    'dc': 'District of Columbia',
    'fl': 'Florida',
    'ga': 'Georgia',
    'hi': 'Hawaii',
    'ia': 'Iowa',
    'il': 'Illinois',
    'in': 'Indiana',
    'ks': 'Kansas',
    'la': 'Louisiana',
    'mi': 'Michigan',
    'mn': 'Minnesota',
    'mo': 'Missouri',
    'nv': 'Nevada',
    'ny': 'New York',
    'oh': 'Ohio',
    'pa': 'Pennsylvania',
    'ri': 'Rhode Island',
    'sc': 'South Carolina',
    'tn': 'Tennessee',
    'tx': 'Texas',
    'ut': 'Utah',
    'va': 'Virginia',
    'wi': 'Wisconsin',
    'wv': 'West Virginia'
}

df_clean['State'] = (
    state_key
    .map(state_map)
    .fillna(state_key.str.title())
)

print(df_clean['State'].value_counts(dropna=False))
```

# 22.Membersihkan Kolom Channel

Menyamakan beberapa bentuk penulisan seperti Call-Center, CALL CENTER, Chat Bot, Chat-bot, E-mail, dan Website.

```python
channel_key = (
    df_clean['Channel']
    .astype('string')
    .str.lower()
    .str.replace(r'[\s-]+', ' ', regex=True)
    .str.strip()
)

channel_map = {
    'call center': 'Call Center',
    'chatbot': 'Chatbot',
    'chat bot': 'Chatbot',
    'email': 'Email',
    'e mail': 'Email',
    'web': 'Web',
    'website': 'Web'
}

df_clean['Channel'] = channel_key.map(channel_map)

print(df_clean['Channel'].value_counts(dropna=False))
```

# 23.Membersihkan Kolom Response Time

Menyamakan penulisan Within sla, WITHIN SLA, dan bentuk lainnya menjadi format yang sama.

```python
response_key = (
    df_clean['Response Time']
    .astype('string')
    .str.lower()
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)

response_map = {
    'within sla': 'Within SLA',
    'above sla': 'Above SLA',
    'below sla': 'Below SLA'
}

df_clean['Response Time'] = response_key.map(response_map)

print(df_clean['Response Time'].value_counts(dropna=False))
```

# 24.Membersihkan Kolom Call Center

Menyamakan format penulisan lokasi call center seperti Los Angeles / CA, Los Angeles, CA, Los Angeles-CA, dan Los Angeles/CA.

```python
call_center_key = (
    df_clean['Call Center']
    .astype('string')
    .str.lower()
    .str.replace(r'\s*[/,-]\s*', '/', regex=True)
    .str.strip()
)

call_center_map = {
    'los angeles/ca': 'Los Angeles/CA',
    'baltimore/md': 'Baltimore/MD',
    'chicago/il': 'Chicago/IL',
    'denver/co': 'Denver/CO'
}

df_clean['Call Center'] = call_center_key.map(call_center_map)

print(df_clean['Call Center'].value_counts(dropna=False))
```

# 25.Membersihkan CSAT Score

CSAT Score harus berupa angka. Pada tahap ini data diubah menjadi tipe numerik dan nilai di luar rentang 1–10 dianggap tidak valid.

```python
df_clean['CSAT Score'] = pd.to_numeric(
    df_clean['CSAT Score'],
    errors='coerce'
)

df_clean.loc[
    ~df_clean['CSAT Score'].between(1, 10),
    'CSAT Score'
] = np.nan

print(df_clean['CSAT Score'].describe())
```

# 26.Membersihkan Call Duration

Durasi panggilan masih memiliki variasi seperti 23, 23 min, 25 min, dan sebagainya. Angka diekstrak sehingga seluruh data menjadi numerik.

```python
df_clean['Call Duration (Minutes)'] = (
    df_clean['Call Duration (Minutes)']
    .astype('string')
    .str.extract(r'(\d+(?:\.\d+)?)', expand=False)
)

df_clean['Call Duration (Minutes)'] = pd.to_numeric(
    df_clean['Call Duration (Minutes)'],
    errors='coerce'
)

print(df_clean['Call Duration (Minutes)'].head(10))
```

# 27.Membersihkan Format Call Timestamp

Tanggal pada dataset memiliki beberapa format berbeda, misalnya January 20, 2024, 08 Jan 2024, 2024-01-04, 01/24/2024, dan 29-01-2024. Semua diubah menjadi format datetime.

```python
df_clean['Call Timestamp'] = pd.to_datetime(
    df_clean['Call Timestamp'],
    errors='coerce',
    format='mixed'
)

print(df_clean['Call Timestamp'].head(10))
print("\nTipe data:", df_clean['Call Timestamp'].dtype)
```

# 28.Mengecek Tanggal yang Gagal Dikonversi

Memastikan tidak ada tanggal yang gagal diproses menjadi format datetime.

```python
invalid_date = df_clean['Call Timestamp'].isna().sum()

print("Jumlah tanggal yang gagal dikonversi:", invalid_date)
```

# 29.Menghapus Data Duplikat Berdasarkan ID

Karena ID merupakan identitas unik setiap data panggilan, data dengan ID yang sama dipertahankan satu saja.

```python
df_clean = (
    df_clean
    .drop_duplicates(subset='ID', keep='first')
    .reset_index(drop=True)
)

print("Jumlah data setelah menghapus duplikat:", len(df_clean))
print("Jumlah ID duplikat:", df_clean['ID'].duplicated().sum())
```

# 30.Menangani Missing Value

Missing value ditangani berdasarkan tipe data. Data kategori diberi label Unknown, sedangkan data numerik menggunakan median.

```python
categorical_columns = [
    'Customer Name',
    'Sentiment',
    'City',
    'State',
    'Channel',
    'Response Time'
]

for col in categorical_columns:
    df_clean[col] = df_clean[col].fillna('Unknown')

df_clean['CSAT Score'] = df_clean['CSAT Score'].fillna(
    df_clean['CSAT Score'].median()
)

df_clean['Call Duration (Minutes)'] = (
    df_clean['Call Duration (Minutes)']
    .fillna(
        df_clean['Call Duration (Minutes)'].median()
    )
)

print("Missing value telah ditangani.")
```

# 31.Mengubah Tipe Data Numerik

Memastikan CSAT dan durasi panggilan mempunyai tipe data numerik yang sesuai.

```python
df_clean['CSAT Score'] = df_clean['CSAT Score'].astype('Int64')

df_clean['Call Duration (Minutes)'] = (
    df_clean['Call Duration (Minutes)']
    .astype('Int64')
)

print(df_clean.dtypes)
```

# 32.Pemeriksaan Outlier Call Duration

IQR digunakan untuk mendeteksi nilai durasi panggilan yang jauh dari distribusi data. Pada tahap ini data belum langsung dihapus.

```python
Q1 = df_clean['Call Duration (Minutes)'].quantile(0.25)
Q3 = df_clean['Call Duration (Minutes)'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

outlier_duration = df_clean[
    (df_clean['Call Duration (Minutes)'] < lower_bound) |
    (df_clean['Call Duration (Minutes)'] > upper_bound)
]

print("Q1            :", Q1)
print("Q3            :", Q3)
print("IQR           :", IQR)
print("Batas bawah   :", lower_bound)
print("Batas atas    :", upper_bound)
print("Jumlah outlier:", len(outlier_duration))
```
